```python
import tkinter as tk
from tkinter import ttk, messagebox
import requests
import threading


# =========================================================
# API Settings
# =========================================================

API_URL = "https://api.one-api.ir/translate/v1/google/"

# توکن جدید خودت را اینجا قرار بده
API_TOKEN = "484758:6abe7ae621551"


# =========================================================
# Application
# =========================================================

class TranslatorApp:

    def __init__(self, root):

        self.root = root

        self.root.title("مترجم آنلاین")
        self.root.geometry("900x750")
        self.root.minsize(650, 550)

        # ---------------- Colors ----------------

        self.bg_color = "#f5f7fb"
        self.card_color = "#ffffff"

        self.primary_color = "#4f46e5"
        self.primary_hover = "#4338ca"

        self.text_color = "#1f2937"
        self.secondary_text = "#6b7280"

        self.border_color = "#e5e7eb"

        self.success_color = "#16a34a"
        self.error_color = "#dc2626"

        self.root.configure(
            bg=self.bg_color
        )

        self.setup_style()
        self.create_scrollable_window()

    # =====================================================
    # Style
    # =====================================================

    def setup_style(self):

        style = ttk.Style()

        try:
            style.theme_use("clam")
        except tk.TclError:
            pass

        style.configure(
            "TCombobox",
            padding=8,
            font=("Tahoma", 10)
        )

    # =====================================================
    # Scrollable Window
    # =====================================================

    def create_scrollable_window(self):

        container = tk.Frame(
            self.root,
            bg=self.bg_color
        )

        container.pack(
            fill="both",
            expand=True
        )

        # Canvas
        self.canvas = tk.Canvas(
            container,
            bg=self.bg_color,
            highlightthickness=0
        )

        self.canvas.pack(
            side="left",
            fill="both",
            expand=True
        )

        # Main scrollbar
        scrollbar = ttk.Scrollbar(
            container,
            orient="vertical",
            command=self.canvas.yview
        )

        scrollbar.pack(
            side="right",
            fill="y"
        )

        self.canvas.configure(
            yscrollcommand=scrollbar.set
        )

        # Main frame
        self.main_frame = tk.Frame(
            self.canvas,
            bg=self.bg_color
        )

        self.canvas_window = self.canvas.create_window(
            (0, 0),
            window=self.main_frame,
            anchor="nw"
        )

        # Update scroll area
        self.main_frame.bind(
            "<Configure>",
            self.update_scroll_region
        )

        # Make frame width equal to canvas
        self.canvas.bind(
            "<Configure>",
            self.resize_canvas
        )

        # Mouse wheel - Windows
        self.canvas.bind_all(
            "<MouseWheel>",
            self.mouse_wheel
        )

        # Mouse wheel - Linux
        self.canvas.bind_all(
            "<Button-4>",
            self.mouse_wheel_linux
        )

        self.canvas.bind_all(
            "<Button-5>",
            self.mouse_wheel_linux
        )

        self.create_widgets()

    # =====================================================
    # Scroll Functions
    # =====================================================

    def update_scroll_region(self, event=None):

        self.canvas.configure(
            scrollregion=self.canvas.bbox("all")
        )

    def resize_canvas(self, event):

        self.canvas.itemconfig(
            self.canvas_window,
            width=event.width
        )

    def mouse_wheel(self, event):

        self.canvas.yview_scroll(
            int(-1 * (event.delta / 120)),
            "units"
        )

    def mouse_wheel_linux(self, event):

        if event.num == 4:

            self.canvas.yview_scroll(
                -3,
                "units"
            )

        elif event.num == 5:

            self.canvas.yview_scroll(
                3,
                "units"
            )

    # =====================================================
    # GUI
    # =====================================================

    def create_widgets(self):

        # =================================================
        # Header
        # =================================================

        header = tk.Frame(
            self.main_frame,
            bg=self.bg_color
        )

        header.pack(
            fill="x",
            padx=30,
            pady=(25, 10)
        )

        tk.Label(
            header,
            text="مترجم آنلاین",
            font=("Tahoma", 24, "bold"),
            bg=self.bg_color,
            fg=self.text_color
        ).pack(
            anchor="e"
        )

        tk.Label(
            header,
            text="ترجمه سریع متن با استفاده از One API",
            font=("Tahoma", 10),
            bg=self.bg_color,
            fg=self.secondary_text
        ).pack(
            anchor="e",
            pady=(5, 0)
        )

        # =================================================
        # API Token
        # =================================================

        token_card = self.create_card()

        tk.Label(
            token_card,
            text="توکن API",
            font=("Tahoma", 10, "bold"),
            bg=self.card_color,
            fg=self.text_color
        ).pack(
            anchor="e",
            padx=15,
            pady=(15, 6)
        )

        self.token_entry = tk.Entry(
            token_card,
            font=("Tahoma", 10),
            show="*",
            relief="flat",
            bg="#f9fafb",
            fg=self.text_color,
            insertbackground=self.text_color
        )

        self.token_entry.pack(
            fill="x",
            padx=15,
            pady=(0, 15),
            ipady=9
        )

        # اگر توکن در کد قرار داده شده باشد
        if API_TOKEN != "YOUR_API_TOKEN":
            self.token_entry.insert(
                0,
                API_TOKEN
            )

        # =================================================
        # Language
        # =================================================

        language_card = self.create_card()

        language_frame = tk.Frame(
            language_card,
            bg=self.card_color
        )

        language_frame.pack(
            fill="x",
            padx=15,
            pady=15
        )

        languages = [
            "fa",
            "en",
            "ar",
            "de",
            "fr",
            "es",
            "tr",
            "ru",
            "it",
            "zh-CN",
            "ja"
        ]

        # Source
        source_frame = tk.Frame(
            language_frame,
            bg=self.card_color
        )

        source_frame.pack(
            side="right",
            padx=10
        )

        tk.Label(
            source_frame,
            text="زبان مبدأ",
            font=("Tahoma", 9),
            bg=self.card_color,
            fg=self.secondary_text
        ).pack(
            anchor="e"
        )

        self.source_var = tk.StringVar(
            value="en"
        )

        self.source_combo = ttk.Combobox(
            source_frame,
            textvariable=self.source_var,
            values=languages,
            state="readonly",
            width=15
        )

        self.source_combo.pack(
            pady=(5, 0)
        )

        # Target
        target_frame = tk.Frame(
            language_frame,
            bg=self.card_color
        )

        target_frame.pack(
            side="right",
            padx=10
        )

        tk.Label(
            target_frame,
            text="زبان مقصد",
            font=("Tahoma", 9),
            bg=self.card_color,
            fg=self.secondary_text
        ).pack(
            anchor="e"
        )

        self.target_var = tk.StringVar(
            value="fa"
        )

        self.target_combo = ttk.Combobox(
            target_frame,
            textvariable=self.target_var,
            values=languages,
            state="readonly",
            width=15
        )

        self.target_combo.pack(
            pady=(5, 0)
        )

        # Swap button
        swap_button = tk.Button(
            language_frame,
            text="⇄",
            font=("Tahoma", 18, "bold"),
            bg=self.card_color,
            fg=self.primary_color,
            activebackground=self.card_color,
            activeforeground=self.primary_hover,
            relief="flat",
            cursor="hand2",
            command=self.swap_languages
        )

        swap_button.pack(
            side="right",
            padx=20
        )

        # =================================================
        # Input
        # =================================================

        input_card = self.create_card()

        tk.Label(
            input_card,
            text="متن ورودی",
            font=("Tahoma", 10, "bold"),
            bg=self.card_color,
            fg=self.text_color
        ).pack(
            anchor="e",
            padx=15,
            pady=(15, 7)
        )

        input_container = tk.Frame(
            input_card,
            bg=self.card_color
        )

        input_container.pack(
            fill="both",
            padx=15,
            pady=(0, 15)
        )

        self.input_text = tk.Text(
            input_container,
            height=10,
            font=("Tahoma", 11),
            wrap="word",
            relief="flat",
            bg="#f9fafb",
            fg=self.text_color,
            insertbackground=self.text_color
        )

        self.input_text.pack(
            side="left",
            fill="both",
            expand=True
        )

        input_scroll = ttk.Scrollbar(
            input_container,
            orient="vertical",
            command=self.input_text.yview
        )

        input_scroll.pack(
            side="right",
            fill="y"
        )

        self.input_text.configure(
            yscrollcommand=input_scroll.set
        )

        self.input_text.insert(
            "1.0",
            "my name is ehsan"
        )

        # Character counter
        self.input_text.bind(
            "<KeyRelease>",
            self.update_character_count
        )

        self.input_count_label = tk.Label(
            input_card,
            text="0 کاراکتر",
            font=("Tahoma", 8),
            bg=self.card_color,
            fg=self.secondary_text
        )

        self.input_count_label.pack(
            anchor="e",
            padx=15,
            pady=(0, 8)
        )

        self.update_character_count()

        # =================================================
        # Output
        # =================================================

        output_card = self.create_card()

        tk.Label(
            output_card,
            text="نتیجه ترجمه",
            font=("Tahoma", 10, "bold"),
            bg=self.card_color,
            fg=self.text_color
        ).pack(
            anchor="e",
            padx=15,
            pady=(15, 7)
        )

        output_container = tk.Frame(
            output_card,
            bg=self.card_color
        )

        output_container.pack(
            fill="both",
            padx=15,
            pady=(0, 15)
        )

        self.output_text = tk.Text(
            output_container,
            height=10,
            font=("Tahoma", 11),
            wrap="word",
            relief="flat",
            bg="#f3f4f6",
            fg=self.text_color,
            state="disabled"
        )

        self.output_text.pack(
            side="left",
            fill="both",
            expand=True
        )

        output_scroll = ttk.Scrollbar(
            output_container,
            orient="vertical",
            command=self.output_text.yview
        )

        output_scroll.pack(
            side="right",
            fill="y"
        )

        self.output_text.configure(
            yscrollcommand=output_scroll.set
        )

        # =================================================
        # Buttons
        # =================================================

        button_frame = tk.Frame(
            self.main_frame,
            bg=self.bg_color
        )

        button_frame.pack(
            fill="x",
            padx=30,
            pady=(5, 10)
        )

        self.translate_button = tk.Button(
            button_frame,
            text="ترجمه",
            font=("Tahoma", 11, "bold"),
            bg=self.primary_color,
            fg="white",
            activebackground=self.primary_hover,
            activeforeground="white",
            relief="flat",
            cursor="hand2",
            command=self.start_translation
        )

        self.translate_button.pack(
            side="right",
            ipadx=30,
            ipady=9
        )

        copy_button = tk.Button(
            button_frame,
            text="کپی نتیجه",
            font=("Tahoma", 10),
            bg="white",
            fg=self.text_color,
            activebackground="#eeeeee",
            relief="flat",
            cursor="hand2",
            command=self.copy_result
        )

        copy_button.pack(
            side="right",
            padx=(0, 10),
            ipadx=12,
            ipady=9
        )

        clear_button = tk.Button(
            button_frame,
            text="پاک کردن",
            font=("Tahoma", 10),
            bg="white",
            fg=self.text_color,
            activebackground="#eeeeee",
            relief="flat",
            cursor="hand2",
            command=self.clear_all
        )

        clear_button.pack(
            side="left",
            ipadx=12,
            ipady=9
        )

        # =================================================
        # Status
        # =================================================

        self.status_label = tk.Label(
            self.main_frame,
            text="آماده",
            font=("Tahoma", 9),
            bg=self.bg_color,
            fg=self.secondary_text
        )

        self.status_label.pack(
            pady=(0, 20)
        )

    # =====================================================
    # Card
    # =====================================================

    def create_card(self):

        card = tk.Frame(
            self.main_frame,
            bg=self.card_color,
            highlightbackground=self.border_color,
            highlightthickness=1
        )

        card.pack(
            fill="x",
            padx=30,
            pady=8
        )

        return card

    # =====================================================
    # Character Counter
    # =====================================================

    def update_character_count(self, event=None):

        text = self.input_text.get(
            "1.0",
            "end-1c"
        )

        self.input_count_label.config(
            text=f"{len(text)} کاراکتر"
        )

    # =====================================================
    # Swap Languages
    # =====================================================

    def swap_languages(self):

        source = self.source_var.get()
        target = self.target_var.get()

        self.source_var.set(target)
        self.target_var.set(source)

        # متن ورودی و خروجی هم جابه‌جا نمی‌شوند؛
        # فقط زبان‌ها تغییر می‌کنند.

    # =====================================================
    # Start Translation
    # =====================================================

    def start_translation(self):

        token = self.token_entry.get().strip()

        text = self.input_text.get(
            "1.0",
            "end-1c"
        ).strip()

        source = self.source_var.get()
        target = self.target_var.get()

        # -----------------------------------------------
        # Validation
        # -----------------------------------------------

        if not token:

            messagebox.showwarning(
                "توکن API",
                "لطفاً توکن API را وارد کنید."
            )

            return

        if not text:

            messagebox.showwarning(
                "متن خالی",
                "لطفاً متنی برای ترجمه وارد کنید."
            )

            return

        if source == target:

            messagebox.showwarning(
                "زبان‌ها",
                "زبان مبدأ و مقصد نمی‌توانند یکسان باشند."
            )

            return

        # -----------------------------------------------
        # UI State
        # -----------------------------------------------

        self.translate_button.config(
            state="disabled",
            text="در حال ترجمه..."
        )

        self.status_label.config(
            text="در حال ارسال درخواست به API...",
            fg=self.secondary_text
        )

        # -----------------------------------------------
        # Thread
        # -----------------------------------------------

        thread = threading.Thread(
            target=self.translate,
            args=(
                token,
                source,
                target,
                text
            ),
            daemon=True
        )

        thread.start()

    # =====================================================
    # API Request
    # =====================================================

    def translate(
        self,
        token,
        source,
        target,
        text
    ):

        headers = {
            "one-api-token": token,
            "Content-Type": "application/json"
        }

        data = {
            "source": source,
            "target": target,
            "text": text
        }

        try:

            response = requests.post(
                API_URL,
                headers=headers,
                json=data,

                # 15 sec connection timeout
                # 90 sec response timeout
                timeout=(15, 90)
            )

            # Debug
            print(
                "STATUS:",
                response.status_code
            )

            print(
                "RESPONSE:",
                response.text
            )

            # -----------------------------------------
            # HTTP Error
            # -----------------------------------------

            response.raise_for_status()

            # -----------------------------------------
            # JSON
            # -----------------------------------------

            try:

                result = response.json()

            except ValueError:

                self.root.after(
                    0,
                    self.show_error,
                    "پاسخ API فرمت JSON معتبر ندارد."
                )

                return

            # -----------------------------------------
            # API Status
            # -----------------------------------------

            if result.get("status") == 200:

                translated_text = (
                    self.extract_translation(result)
                )

                self.root.after(
                    0,
                    self.show_result,
                    translated_text
                )

            else:

                error_message = result.get(
                    "message",
                    result.get(
                        "error",
                        "API درخواست را قبول نکرد."
                    )
                )

                self.root.after(
                    0,
                    self.show_error,
                    f"خطای API:\n\n{error_message}"
                )

        # =================================================
        # Connection Timeout
        # =================================================

        except requests.exceptions.ConnectTimeout:

            self.root.after(
                0,
                self.show_error,
                "اتصال به سرور API در زمان تعیین‌شده برقرار نشد."
            )

        # =================================================
        # Read Timeout
        # =================================================

        except requests.exceptions.ReadTimeout:

            self.root.after(
                0,
                self.show_error,
                "سرور API در زمان تعیین‌شده پاسخ نداد.\n\n"
                "ممکن است متن بسیار طولانی باشد یا سرور "
                "در حال پردازش درخواست باشد."
            )

        # =================================================
        # Connection Error
        # =================================================

        except requests.exceptions.ConnectionError as e:

            self.root.after(
                0,
                self.show_error,
                f"خطا در اتصال به API:\n\n{e}"
            )

        # =================================================
        # HTTP Error
        # =================================================

        except requests.exceptions.HTTPError:

            self.root.after(
                0,
                self.show_error,
                f"خطای HTTP:\n\n"
                f"Status Code: {response.status_code}\n\n"
                f"{response.text}"
            )

        # =================================================
        # Other Error
        # =================================================

        except Exception as e:

            self.root.after(
                0,
                self.show_error,
                f"خطای غیرمنتظره:\n\n{e}"
            )

    # =====================================================
    # Extract Translation
    # =====================================================

    def extract_translation(self, result):

        """
        پاسخ واقعی API:

        {
            "status": 200,
            "result": "سلام حالتون چطوره؟"
        }
        """

        if isinstance(result, dict):

            if "result" in result:

                return str(
                    result["result"]
                )

        return str(result)

    # =====================================================
    # Show Result
    # =====================================================

    def show_result(self, result):

        self.output_text.config(
            state="normal"
        )

        self.output_text.delete(
            "1.0",
            "end"
        )

        self.output_text.insert(
            "1.0",
            result
        )

        self.output_text.config(
            state="disabled"
        )

        self.translate_button.config(
            state="normal",
            text="ترجمه"
        )

        self.status_label.config(
            text="ترجمه با موفقیت انجام شد.",
            fg=self.success_color
        )

        # نمایش ابتدای نتیجه
        self.output_text.see(
            "1.0"
        )

        # Update main scroll
        self.root.after(
            100,
            self.update_scroll_region
        )

    # =====================================================
    # Show Error
    # =====================================================

    def show_error(self, error):

        self.translate_button.config(
            state="normal",
            text="ترجمه"
        )

        self.status_label.config(
            text="خطا در انجام درخواست",
            fg=self.error_color
        )

        messagebox.showerror(
            "خطا",
            error
        )

    # =====================================================
    # Copy Result
    # =====================================================

    def copy_result(self):

        result = self.output_text.get(
            "1.0",
            "end-1c"
        ).strip()

        if not result:

            messagebox.showinfo(
                "کپی",
                "هنوز نتیجه‌ای برای کپی وجود ندارد."
            )

            return

        self.root.clipboard_clear()

        self.root.clipboard_append(
            result
        )

        self.status_label.config(
            text="نتیجه در کلیپ‌بورد کپی شد.",
            fg=self.success_color
        )

    # =====================================================
    # Clear
    # =====================================================

    def clear_all(self):

        self.input_text.delete(
            "1.0",
            "end"
        )

        self.output_text.config(
            state="normal"
        )

        self.output_text.delete(
            "1.0",
            "end"
        )

        self.output_text.config(
            state="disabled"
        )

        self.status_label.config(
            text="آماده",
            fg=self.secondary_text
        )

        self.update_character_count()


# =========================================================
# Run
# =========================================================

if __name__ == "__main__":

    root = tk.Tk()

    app = TranslatorApp(root)

    root.mainloop()
```
